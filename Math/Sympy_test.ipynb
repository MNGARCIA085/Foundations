```python
import sympy as sp

# Variables
x1, x2 = sp.symbols("x1 x2")

# c : R² -> R²
c = sp.Matrix([
    x1 + x2,
    x1 * x2
])

# Intermediate variables
v1, v2 = sp.symbols("v1 v2")
u1, u2 = sp.symbols("u1 u2")

# b : R² -> R²
b = sp.Matrix([
    v1**2,
    v1 + v2
])

# a : R² -> R²
a = sp.Matrix([
    u1 + u2,
    u1 * u2
])

# Jacobians
Ja = a.jacobian([u1, u2])
Jb = b.jacobian([v1, v2])
Jc = c.jacobian([x1, x2])

print("a'(u) =")
sp.pprint(Ja)

print("\nb'(v) =")
sp.pprint(Jb)

print("\nc'(x) =")
sp.pprint(Jc)

# Compose the Jacobians symbolically
J = Ja * Jb * Jc

print("\nJ before substitutions =")
sp.pprint(sp.factor(J))

# Substitutions:
subs = {
    v1: c[0],
    v2: c[1],
    u1: b[0].subs({v1: c[0], v2: c[1]}),
    u2: b[1].subs({v1: c[0], v2: c[1]})
}

J_final = sp.simplify(J.subs(subs))

print("\nFinal Jacobian:")
sp.pprint(J_final)
```

```python
import sympy as sp
from sympy.tensor.array import derive_by_array

x, y = sp.symbols('x y')

F = sp.Matrix([
    [x**2, x*y],
    [x + y, y**2]
])

T = derive_by_array(F, (x, y))

print(T)
print(T.shape)
```
